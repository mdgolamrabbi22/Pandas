```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Clean_data.csv")
```

```python
plt.figure(figsize=(10,5))
bar= sns.countplot(data=df,x="City",hue="Customer_Rating",palette="Set1")
for container in bar.containers:
    bar.bar_label(container)
plt.title("Customer Rating Chart",color="#105fde",size=20)
```

```python
plt.figure(figsize=(20,13))
bar = sns.countplot(data=df,x="City",hue="Category",palette="Set2")
for container in bar.containers:
    bar.bar_label(container)
plt.title("Category in City Chart",color="#105fde",size=20)
```

```python
plt.figure(figsize=(10,8))
bar = sns.barplot(data=df, x="Category", y="Qty", palette="Set1",estimator="sum", errorbar=None)
for container in bar.containers:
    bar.bar_label(container)
plt.title("Total Sales Quantity by Category",color="#105fde",size=20)
plt.xlabel("Category",color="#09e68d",size=16)
plt.ylabel("Qty_Count",color="#09e68d",size=16)
```

```python
plt.figure(figsize=(12, 6))
bar = sns.countplot(
    data=df,
    x="City",
    order=df["City"].value_counts().index
)
for container in bar.containers:
    bar.bar_label(container, padding=3)

plt.title("City-wise Customer Distribution")
plt.xlabel("City")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
bins = [-0.01, 5, 10, 20, 30, 100]

labels = ["0-5%", "6-10%", "11-20%", "21-30%", "30%+"]

df["Discount_Group"] = pd.cut(
    df["Discount_Pct"],
    bins=bins,
    labels=labels
)
plt.figure(figsize=(12, 6))
bar = sns.barplot(
    data=df,
    x="Discount_Group",
    y="Qty",
    estimator="sum",
    errorbar=None
)
for container in bar.containers:
    bar.bar_label(container)
plt.title("Discount wise Qty",color="#105fde",size=20)
plt.xlabel("Discount percentise",color="#09e68d",size=16)
plt.ylabel("Count",color="#09e68d",size=16)
```
