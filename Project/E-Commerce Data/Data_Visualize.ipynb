```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Clean_data.csv")
```

```python
df.head()
```

```python
df[df["Customer_Id"].isnull()]
```

```python
df.columns
```

```python
plt.figure(figsize=(10,5))
bar= sns.countplot(data=df,x="City",hue="Customer_Rating",palette="Set1")
for container in bar.containers:
    bar.bar_label(container)
plt.title("Customer Rating Chart",color="#105fde",size=20)
```

```python
plt.figure(figsize=(20,13))
bar = sns.countplot(data=df,x="City",hue="Category",palette="Set2")
for container in bar.containers:
    bar.bar_label(container)
plt.title("Category in City Chart",color="#105fde",size=20)
```

```python
plt.figure(figsize=(10,8))
bar = sns.barplot(data=df, x="Category", y="Qty", palette="Set1",estimator="sum", errorbar=None)
for container in bar.containers:
    bar.bar_label(container)
plt.title("Total Sales Quantity by Category",color="#105fde",size=20)
plt.xlabel("Category",color="#09e68d",size=16)
plt.xlabel("Category",color="#09e68d",size=16)
plt.ylabel("Qty_Count",color="#09e68d",size=16)
```
