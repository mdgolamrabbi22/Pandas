```python
df["Customer_ID"] = df["Customer_ID"].fillna("None")
df["Customer_ID"].isnull().sum()
```

```python
df.columns = df.columns.str.strip().str.title().str.replace(" ","_")
df.columns
```

```python
df["Age"].isnull().sum()
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Age"].dtype
```

```python
df["Age"].isnull().sum()
```

```python
mean = df["Age"].mean()
std = df["Age"].std()
zscore = (df["Age"] - mean) / std
zscore
```

```python
df["Zscore"] = (df["Age"] - mean) / std
```

```python
df.loc[
    (df["Zscore"] > 3) | (df["Zscore"] < -3),
    "Age"
] = np.nan


```

```python
df["Age"] =df["Age"].fillna(mean)
df["Age"].isnull().sum()
```

```python
df.isnull().sum()
```

```python
df["Gender"] = (df["Gender"].str.strip().str.title().replace({
    "M" : "Male",
    "F" : "Female",
    "?" : np.nan,
    "Missing" : np.nan,
    "-" : np.nan
}))
```

```python
df["Gender"].unique()
```

```python
df["Gender"] = df["Gender"].fillna("Unknown")
```
