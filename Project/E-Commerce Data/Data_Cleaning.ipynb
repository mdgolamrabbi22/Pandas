```python
def clean_money(col):
    return pd.to_numeric(col.astype(str).str.replace(r'[^0-9.\-]', '', regex=True), errors='coerce')

df["Unit_Price"] = clean_money(df["Unit_Price"])
df["Total_Amount"] = clean_money(df["Total_Amount"])
```

```python
df.isnull().sum()
```

```python
calc = df["Unit_Price"] * df["Qty"] * (1 - df["Discount_Pct"] / 100)
df["Total_Amount"] = df["Total_Amount"].fillna(calc).round(2)
df["Total_Amount"].isnull().sum()
```

```python
disc_f = 1 - df["Discount_Pct"] / 100

# Jekhane Unit_Price null, baki 3 ta ache
m = (df["Unit_Price"].isnull()
     & df["Total_Amount"].notnull()
     & df["Qty"].notnull()
     & df["Discount_Pct"].notnull())

df.loc[m, "Unit_Price"] = (df.loc[m, "Total_Amount"] / (df.loc[m, "Qty"] * disc_f[m])).round(2)

print(df["Unit_Price"].isnull().sum())   # koto ta baki
```

```python
df.isnull().sum()
```

```python
# Bhul value NaN
df.loc[(df['Unit_Price'] <= 0) | (df['Unit_Price'] > 1_000_000), 'Unit_Price'] = np.nan
df.loc[(df['Qty'] <= 0) | (df['Qty'] >= 100), 'Qty'] = np.nan
df.loc[~df['Discount_Pct'].between(0, 100), 'Discount_Pct'] = np.nan
 
# 6. Total_Amount theke ulto hishab
disc_f = 1 - df['Discount_Pct'] / 100
 
# Unit_Price = Total / (Qty x discount factor)
m = df['Unit_Price'].isna() & df['Qty'].notna() & df['Discount_Pct'].notna() & df['Total_Amount'].notna()
df.loc[m, 'Unit_Price'] = (df.loc[m, 'Total_Amount'] / (df.loc[m, 'Qty'] * disc_f[m])).round(2)
 
# Qty = Total / (Unit_Price x discount factor)
m = df['Qty'].isna() & df['Unit_Price'].notna() & df['Discount_Pct'].notna() & df['Total_Amount'].notna()
q = df.loc[m, 'Total_Amount'] / (df.loc[m, 'Unit_Price'] * disc_f[m])
ok = (q - q.round()).abs() < 0.05
df.loc[q[ok].index, 'Qty'] = q[ok].round()
 
# Discount = (1 - Total / (Unit_Price x Qty)) x 100, tarpor kache r valid discount e snap
valid_disc = np.array([0, 5, 10, 15, 20, 25])
m = df['Discount_Pct'].isna() & df['Unit_Price'].notna() & df['Qty'].notna() & df['Total_Amount'].notna()
d = ((1 - df.loc[m, 'Total_Amount'] / (df.loc[m, 'Unit_Price'] * df.loc[m, 'Qty'])) * 100).astype(float)
snapped = pd.Series(valid_disc[np.abs(d.to_numpy()[:, None] - valid_disc).argmin(axis=1)], index=d.index)
ok = (d - snapped).abs() < 1.5
df.loc[snapped[ok].index, 'Discount_Pct'] = snapped[ok]
```
