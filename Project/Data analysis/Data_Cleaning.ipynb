```python
df = df.drop_duplicates(subset="Order_Id").reset_index(drop=True)
```

```python
mean = df["Unit_Price"].mean()
std = df["Unit_Price"].std()
zscore = (df["Unit_Price"] - mean) / std 
```

```python
df.loc[(zscore > 3 ) | (zscore < -3 ),"Unit_Price"] = np.nan
```

```python
df["Unit_Price"].isnull().sum()
```

```python
mask = df["Unit_Price"].isnull()

df.loc[mask, "Unit_Price"] = (df.loc[mask, "Total_Amount"]
    / (df.loc[mask, "Quantity"] * (1 - df.loc[mask, "Discount_Pct"] / 100))
).round(0)
```
