```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df =pd.read_csv("Clean_Data.csv")
```

```python
df.columns
```

```python
plt.figure(figsize=(10,8))
bar = sns.countplot(data=df,x="Gender",palette="Set1")
for container in bar.containers:
    bar.bar_label(container)
plt.title("Gender Chart",size=20,color="purple")
plt.xlabel("Sex",size=20,color="purple")
plt.ylabel("Count",size=20,color="purple")
```

**Gender** = Gender daikha buja jai akahne Male Basi and Female kom

```python
plt.figure(figsize=(10,8))
sort = df["City"].value_counts().index
bar = sns.countplot(data =  df,x="City",order=sort, palette="Set1")
for container in bar.containers:
    bar.bar_label(container)
plt.title("City Chart",size=20,color="purple")
plt.xlabel("City",size=20,color="purple")
plt.ylabel("Count",size=20,color="purple")
```
