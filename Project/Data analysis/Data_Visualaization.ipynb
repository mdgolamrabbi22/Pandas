```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df =pd.read_csv("Clean_Data.csv")
```

```python
plt.figure(figsize=(10,8))
bar = sns.countplot(data=df,x="Gender",palette="Set1")
for container in bar.containers:
    bar.bar_label(container)
plt.title("Gender Chart",size=20,color="purple")
plt.xlabel("Sex",size=20,color="purple")
plt.ylabel("Count",size=20,color="purple")
```

```python
plt.figure(figsize=(10,8))
sort = df["City"].value_counts().index
bar = sns.countplot(data =  df,x="City",order=sort, palette="Set1")
for container in bar.containers:
    bar.bar_label(container)
plt.title("City Chart",size=20,color="purple")
plt.xlabel("City",size=20,color="purple")
plt.ylabel("Count",size=20,color="purple")
```

```python
plt.figure(figsize=(20,12))
bar = sns.countplot(data=df,x="City",hue="Category",palette="Set1")
for container in bar.containers:
    bar.bar_label(container, fontsize=15, fontweight="bold")
plt.title("City Chart",fontsize=40,color="purple")
plt.xlabel("City",fontsize=40,color="purple")
plt.ylabel("Count",size=40,color="purple")
plt.xticks(fontsize=18)
plt.yticks(fontsize=18)
plt.legend(title="Category", title_fontsize=30, fontsize=20, bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
```

### 📊 Insights: City-wise Product Category Distribution

* **Top Performing Market (Dhaka):** Dhaka generates the highest number of orders across all categories. **Books (33 orders)** is the single most popular category in Dhaka, followed closely by **Clothing (29)** and **Beauty (28)**.
* **Second Largest Market (Chattogram):** Chattogram shows strong demand across every category, led by **Electronics (22 orders)** and **Grocery (21 orders)**.
* **Regional Preferences:** 
  * In **Sylhet**, **Home (13)** and **Clothing (13)** are the top choices, while **Beauty (2)** has the lowest demand.
  * In **Khulna**, **Grocery (14)** and **Books (13)** lead the sales.
  * In **Rajshahi**, **Beauty (9)** and **Clothing (9)** are equally popular.
* **Low Order Volume Areas:** **Barishal** and **Rangpur** have comparatively lower order volumes across all categories, indicating a need for targeted marketing campaigns in these regions.
* **Data Quality Note:** A small number of orders are recorded under **Unknown_City**, which may require better address validation during checkout.
